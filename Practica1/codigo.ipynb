#### 1. Importamos librerías

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.model_selection import KFold
from sklearn.model_selection import LeaveOneOut
from sklearn.utils import resample
```

#### 2. Utilizamos un dataframe de pandas para leer el archivo .csv 
Luego, definimos la característica que será el input, y la columna objetivo

```python
dataset = pd.read_csv('metodosDeValidacion.csv')

X =dataset['x']
y = dataset['y']

print("Datos originales:")
for i in range(len(X)):
    print(f'X: {X.iloc[i]}, y: {y.iloc[i]}')
```

---
### 3. Dividimos el dataset en conjunto de entrenamiento y prueba

Usando la función **sklearn.model_selection.train_test_split(*arrays, test_size=None, train_size=None, random_state=None, shuffle=True, stratify=None)**


* random_state: Un entero para la reproducibilidad. Al configurarlo, se garantiza que se genere la misma división cada vez que se ejecuta el código.
* shuffle: (El valor predeterminado es Verdadero) Determina si se deben mezclar los datos antes de dividirlos.
* stratify: (Opcional) Se utiliza para muestreo estratificado, especialmente útil con conjuntos de datos desequilibrados. Si se proporciona con "y", garantiza que la proporción de clases en los conjuntos de entrenamiento y prueba sea similar a la del conjunto de datos original.


```python

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, shuffle=False)
```

```python
print("\nDatos de entrenamiento:")
for i in range(len(X_train)):
    print(f'x: {X_train.iloc[i]} | y: {y_train.iloc[i]}')
    
print("\nDatos de prueba:")
for i in range(len(X_test)):
    print(f'x: {X_test.iloc[i]} | y: {y_test.iloc[i]}')
```

---
### 4. Validación Cruzada: 6 folds
Usando la función **sklearn.model_selection.KFold**
* n_splits (k): El número de pliegues a crear. Debe ser al menos 2.
* shuffle: Un valor booleano que indica si se deben mezclar los datos antes de dividirlos en pliegues. El valor predeterminado es Falso
* random_state: Se utiliza junto con shuffle=True para garantizar la reproducibilidad del mezclado. Al establecerlo en un entero, se producirá la misma división en cada ocasión.

```python
kf = KFold(n_splits=6, shuffle=False)

X = X_train
y = y_train
```

```python

for fold, (ind_entrenam, ind_prueba) in enumerate(kf.split(X)):
    print(f"\nFold {fold + 1}")

    print("Datos de entrenamiento:")
    for i in ind_entrenam:
        print(f'x: {X.iloc[i]}  | y: {y.iloc[i]}')

    print("Datos de prueba:")
    for i in ind_prueba:
        print(f'x: {X.iloc[i]}  | y: {y.iloc[i]}')

    print("-" * 20)
```

---
### 5. Conjuntos de Validación con Leave one out
Utilizando **LeaveOneOut()**
* Cada muestra individual del conjunto de datos se utiliza como conjunto de prueba una vez, mientras que las muestras restantes forman el conjunto de entrenamiento.
* k-folds donde k es igual al número de muestras.

```python
loo = LeaveOneOut()
```

```python


for fold, (ind_entrenam, ind_prueba) in enumerate(loo.split(X)):
    print(f"\nFold {fold + 1}")

    print("Datos de entrenamiento:")
    for i in ind_entrenam:
        print(f'x: {X.iloc[i]}  | y: {y.iloc[i]}')

    print("Datos de prueba:")
    for i in ind_prueba:
        print(f'x: {X.iloc[i]}  | y: {y.iloc[i]}')

    print("-" * 20)
```

---
### 6. Utilizando *Bootstrap* generamos 2 conjuntos de validación con conjuntos de entrenamiento de tamaño = 9 con resample().  
**sklearn.utils.resample(*arrays, replace=True, n_samples=None, random_state=None, stratify=None, sample_weight=None)
**


```python
X_train_boot = resample(X, n_samples=9, replace=True, random_state=1)
y_train_boot = resample(y, n_samples=9, replace=True, random_state=1)

print("\nBootstrap Sample 1 - Datos de entrenamiento:")
for i in range(len(X_train_boot)):
    print(f'x: {X_train_boot.iloc[i]}, y: {y_train_boot.iloc[i]}')

print("\nBootstrap Sample 1 - Datos de prueba:")
for i in range(len(X)):
    if X.iloc[i] not in X_train_boot.values:
        print(f'x: {X.iloc[i]}, y: {y.iloc[i]}')
```

```python


X_train_boot = resample(X, n_samples=9, replace=True, random_state=42)
y_train_boot = resample(y, n_samples=9, replace=True, random_state=42)

print("\nBootstrap Sample 2 - Datos de entrenamiento:")
for i in range(len(X_train_boot)):
    print(f'x: {X_train_boot.iloc[i]}, y: {y_train_boot.iloc[i]}')

print("\nBootstrap Sample 2 - Datos de prueba:")
for i in range(len(X)):
    if X.iloc[i] not in X_train_boot.values:
        print(f'x: {X.iloc[i]}, y: {y.iloc[i]}')
```

**sklearn.model_selection.KFold:**
* Divide un dataset en k subconjuntos del mismo tamaño, por default no mezcla los datos.
* Stratificar se refiere a cuando tenemos clases de diferentes tamaños, hace las k pliegues considerando un pliegue prporcional para cada clase.
