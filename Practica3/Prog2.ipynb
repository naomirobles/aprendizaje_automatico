## Especificaciones

### I. Para este programa se utilizará el dataset datos.csv
* Entrada: archivo *datos.csv*, número de iteraciones para SGD y $\alpha$

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import SGDRegressor
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import RobustScaler
import operator
```

```python
# Leer datos
df = pd.read_csv('cal_housing.csv')

X = df.iloc[:, :-1].values
y = df['medianHouseValue'].values
```

### II. Divide *datos.csv* en 70% para entrenamiento y 30% para pruebas
Con los parámetros *shuffle=True* y *random_state=0*

```python

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state = 0, shuffle=True)
```

### III. Con las bibliotecas de scikit-learn realiza las siguientes regresiones
* Regresión polinomial de grado 2
* Regresión polinomial de grado 2 con escalamiento estándar
* Regresión polinomial de grado 2 con escalamiento robusto
* Regresión polinomial de grado 3
* Regresión polinomial de grado 3 con escalamiento estándar
* Regresión polinomial de grado 3 con escalamiento robusto

#### III.a Elevar el grado de los datos

```python

# -----  POLINOMIAL 2do GRADO ------
poly2 = PolynomialFeatures(degree=2)
X_train_poly2 = poly2.fit_transform(X_train)
X_test_poly2 = poly2.transform(X_test)

# -----  POLINOMIAL 3er GRADO ------
poly3 = PolynomialFeatures(degree=3)
X_train_poly3 = poly3.fit_transform(X_train)
X_test_poly3 = poly3.transform(X_test)

# 1. Regresión lineal sin escalado
modelo_lineal = LinearRegression()
modelo_lineal.fit(X_train, y_train)
y_pred_lineal = modelo_lineal.predict(X_test)


```

#### III.b Escalamos los datos elevados

```python



# ------ Escalamiento estandarizado -------------
# Escalar X (Grado 2)
standard_scaler_X = StandardScaler()
X_train_poly2_standard_scaled = standard_scaler_X.fit_transform(X_train_poly2)
X_test_poly2_standard_scaled = standard_scaler_X.transform(X_test_poly2)

# Escalar X (Grado 3)
X_train_poly3_standard_scaled = standard_scaler_X.fit_transform(X_train_poly3)
X_test_poly3_standard_scaled = standard_scaler_X.transform(X_test_poly3)


# Escalar Y 
standard_scaler_y = StandardScaler()
y_train_standard_scaled = standard_scaler_y.fit_transform(y_train.reshape(-1, 1)).ravel()
y_test_standard_scaled = standard_scaler_y.transform(y_test.reshape(-1, 1)).ravel()


# ------ Escalamiento robusto -------------------
# Escalar X (Grado 2)
robust_scaler_X = RobustScaler()
X_train_poly2_robust_scaled = robust_scaler_X.fit_transform(X_train_poly2)
X_test_poly2_robust_scaled = robust_scaler_X.transform(X_test_poly2)

# Escalar X (Grado 3)
X_train_poly3_robust_scaled = robust_scaler_X.fit_transform(X_train_poly3)
X_test_poly3_robust_scaled = robust_scaler_X.transform(X_test_poly3)


# Escalar Y 
robust_scaler_y = RobustScaler()
y_train_robust_scaled = robust_scaler_y.fit_transform(y_train.reshape(-1, 1)).ravel()
y_test_robust_scaled = robust_scaler_y.transform(y_test.reshape(-1, 1)).ravel()



```

```python
# --- Modelos con escalamiento estandarizado ---

# 1. Regresión Polinomeal de grado 2
modelo_poly2 = LinearRegression()
modelo_poly2.fit(X_train_poly2, y_train)
y_pred_poly2_unscaled = modelo_poly2.predict(X_test_poly2)
y_pred_poly2 = y_pred_poly2_unscaled.ravel()

# 2. Regresión Polinomeal de grado 2 con escalado estandarizado
modelo_poly2.fit(X_train_poly2_standard_scaled, y_train_standard_scaled)
y_pred_poly2_standard_scaled = modelo_poly2.predict(X_test_poly2_standard_scaled)
y_pred_poly2_standard_scaled = y_pred_poly2_standard_scaled.ravel()

# 3. Regresión Polinomeal de grado 2 con escalado robusto
modelo_poly2.fit(X_train_poly2_robust_scaled, y_train_robust_scaled)
y_pred_poly2_robust_scaled = modelo_poly2.predict(X_test_poly2_robust_scaled)
y_pred_poly2_robust_scaled = y_pred_poly2_robust_scaled.ravel()

# 4. Regresión Polinomeal de grado 3
modelo_poly3 = LinearRegression()
modelo_poly3.fit(X_train_poly3, y_train)
y_pred_poly3_unscaled = modelo_poly3.predict(X_test_poly3)
y_pred_poly3 = y_pred_poly3_unscaled.ravel()

# 2. Regresión Polinomeal de grado 2 con escalado estandarizado
modelo_poly3.fit(X_train_poly3_standard_scaled, y_train_standard_scaled)
y_pred_poly3_standard_scaled = modelo_poly3.predict(X_test_poly3_standard_scaled)
y_pred_poly3_standard_scaled = y_pred_poly3_standard_scaled.ravel()

# 3. Regresión Polinomeal de grado 2 con escalado robusto
modelo_poly3.fit(X_train_poly3_robust_scaled, y_train_robust_scaled)
y_pred_poly3_robust_scaled = modelo_poly3.predict(X_test_poly3_robust_scaled)
y_pred_poly3_robust_scaled = y_pred_poly3_robust_scaled.ravel()
```

```python
# Desescalar las predicciones de los modelos escalados

# Grado 2 con escalado estándar
y_pred_poly2_standard_unscaled = standard_scaler_y.inverse_transform(
    y_pred_poly2_standard_scaled.reshape(-1, 1)
).ravel()

# Grado 2 con escalado robusto
y_pred_poly2_robust_unscaled = robust_scaler_y.inverse_transform(
    y_pred_poly2_robust_scaled.reshape(-1, 1)
).ravel()

# Grado 3 con escalado estándar
y_pred_poly3_standard_unscaled = standard_scaler_y.inverse_transform(
    y_pred_poly3_standard_scaled.reshape(-1, 1)
).ravel()

# Grado 3 con escalado robusto
y_pred_poly3_robust_unscaled = robust_scaler_y.inverse_transform(
    y_pred_poly3_robust_scaled.reshape(-1, 1)
).ravel()


# Calcular métricas para cada modelo (ahora todos con y_test sin escalar)
modelos = {
        'Regresión lineal': (y_pred_lineal, 
                        mean_squared_error(y_test, y_pred_lineal), 
                        r2_score(y_test, y_pred_lineal)),
    'Regresión polinomial de grado 2': (y_pred_poly2, 
                                        mean_squared_error(y_test, y_pred_poly2), 
                                        r2_score(y_test, y_pred_poly2)),
    
    'Regresión polinomial de grado 2 con escalado estándar': (y_pred_poly2_standard_unscaled, 
                                                            mean_squared_error(y_test, y_pred_poly2_standard_unscaled), 
                                                            r2_score(y_test, y_pred_poly2_standard_unscaled)),
    
    'Regresión polinomial de grado 2 con escalado robusto': (y_pred_poly2_robust_unscaled, 
                                                             mean_squared_error(y_test, y_pred_poly2_robust_unscaled), 
                                                             r2_score(y_test, y_pred_poly2_robust_unscaled)),
    
    'Regresión polinomial de grado 3': (y_pred_poly3, 
                                        mean_squared_error(y_test, y_pred_poly3), 
                                        r2_score(y_test, y_pred_poly3)),
    
    'Regresión polinomial de grado 3 con escalado estándar': (y_pred_poly3_standard_unscaled, 
                                                              mean_squared_error(y_test, y_pred_poly3_standard_unscaled), 
                                                              r2_score(y_test, y_pred_poly3_standard_unscaled)),
    
    'Regresión polinomial de grado 3 con escalado robusto': (y_pred_poly3_robust_unscaled, 
                                                             mean_squared_error(y_test, y_pred_poly3_robust_unscaled), 
                                                             r2_score(y_test, y_pred_poly3_robust_unscaled)),
}

# Resumen de resultados
print("\n" + "="*90)
print(f"{'Modelo':<60} {'MSE':>15} {'R²':>12}")
print("="*90)

for nombre, (_, mse, r2) in modelos.items():
    print(f"{nombre:<60} {mse:>15.4f} {r2:>12.4f}")

print("="*90)
```
